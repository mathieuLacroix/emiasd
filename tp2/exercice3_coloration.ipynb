# Colorier un graphe


Étant donné un graphe $G = (V, E)$, une **_coloration_** de $G$ est une affectation d'une couleur à chaque sommet telle que deux sommets adjacents n'ont pas la même couleur. Le **_nombre chromatique_** de $G$ est le nombre minimal de couleurs nécessaires pour colorier $G$. Le **_problème de coloration_** consiste à déterminer le nombre chromatique d'un graphe. C'est un problème d'optimisation qui modélise plusieurs problèmes de recherche opérationnelle tels que l'affectation de fréquences ou la planification d'examens.
