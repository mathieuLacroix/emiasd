# Colorier un graphe

Étant donné un graphe $G = (V, E)$, une **_coloration_** de $G$ est une affectation d'une couleur à chaque sommet telle que deux sommets adjacents n'ont pas la même couleur. Le **_nombre chromatique_** de $G$ est le nombre minimal de couleurs nécessaires pour colorier $G$. Le **_problème de coloration_** consiste à déterminer le nombre chromatique d'un graphe. C'est un problème d'optimisation qui modélise plusieurs problèmes de recherche opérationnelle tels que l'affectation de fréquences ou la planification d'examens.

## Question 2

Formulez le problème de coloration de graphe comme un problème linéaire en nombres entiers mixtes.

**CORRECTION :**


Soit $G=(V,E)$ un graphe sans sommets isolés. Soit $C$ un ensemble de couleurs contenant assez de couleurs pour colorier $G$.

Soit $x_v^c = \left\{ \begin{array}{ll} 1 & \text{si $v$ est colorié avec la couleur $c$}\\ 0 & \text{sinon}\end{array} \right .~~$ pour tout $v \in V$ et $c \in C$.

Soit $y^c = \left\{ \begin{array}{ll} 1 & \text{si la couleur $c$ est utilisée}\\ 0 & \text{sinon}\end{array} \right .~~$ pour tout $c \in C$. Cependant, les contraintes d'intégrité sur $y$ peuvent être relâchées.


Le problème de coloration peut être formulé comme suit.
$$
\begin{align}
&\min \sum_{c \in C} y^c\\
&\sum_{c \in C} x_v^c  = 1 & \quad & \forall v \in V\\
& x_u^c + x_v^c \le y^c & & \forall uv \in E, \forall c \in C\\
& x_v^c \in \{0, 1\} & & \forall v \in V, \forall c \in C\\
& 0 \le y^c \le 1 & & \forall c \in C\\
\end{align}
$$

## Question 3

Implémentez le PLNE pour trouver le nombre chromatique du graphe donné dans l'exemple.


```python
edges = [
    (0, 1),
    (0, 7),
    (1, 2),
    (1, 7),
    (1, 8),
    (2, 3),
    (2, 8),
    (3, 4),
    (3, 6),
    (4, 5),
    (4, 6),
    (5, 6),
    (6, 7),
    (6, 8),
    (7, 8)
]
n = max(max(e[0], e[1]) for e in edges) + 1
```

```python
##################
#   Correction   #
##################


def coloring(n, edges, max_number_of_colors):
    m = Model()
    C = range(max_number_of_colors)
    V = range(n+1)
    x = [
            [m.add_var(var_type = BINARY) for c in C] for v in V
    ]
    y = [ m.add_var(ub = 1) for c in C]
    m.objective = minimize(xsum(y[c] for c in C))

    # for each vertex
    for v in V:
        m += xsum(x[v][c] for c in C) == 1

    # for each edge
    for e in edges:
        # for each color
        for c in C:
            m += x[e[0]][c] + x[e[1]][c] <= y[c]
    m.verbose = 0 # to remove the output of the solution process
    status = m.optimize()
    print(f"Chromatic number: {round(m.objective_value)}")
    for c in C:
        if round(y[c].x) == 1:
            verticies = [v for v in V if round(x[v][c].x) == 1]
            print(f"Color class: {verticies}")

coloring(n, edges, 4)      
```
