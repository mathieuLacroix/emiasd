# Comment approvisionner les magasins depuis les entrepôts ?
 - Correction
Un producteur dispose de $m$ entrepôts ($i = 1, 2,\dots, m$) et doit approvisionner $n$ magasins ($j = 1, 2,\dots, n$). Chaque entrepôt $i$ peut livrer au plus une quantité $c_i$ de produit par semaine. Chaque magasin $j$ a une demande hebdomadaire connue de $d_j$.
Le coût d'expédition d'une unité de produit de l'entrepôt i vers le magasin j a été estimé à $p_{ij}$ euros par unité.

## Question 1

Écrivez un modèle PL pour minimiser le coût total afin de satisfaire les demandes des magasins.
Identifiez les décisions qui doivent être prises et les variables correspondantes qui doivent être utilisées. Identifiez et commentez la fonction objectif du problème et les contraintes.

**CORRECTION :**


Soient $I$ et $J$ les ensembles des entrepôts et des magasins respectivement.

La forme générale du PL pour résoudre le problème est :
$$
\begin{align}
& \min \sum_{i \in I} \sum_{j \in J} p_{ij} x_{ij}\\
& \sum_{j \in J} x_{ij} \le c_i & & ∀ i \in I\\
& \sum_{i \in I} x_{ij} \ge d_j & & ∀ j \in J\\
& x_{ij} \ge 0 & & ∀ i \in I, j \in J
\end{align}
$$


## Question 2

Considérons maintenant le cas suivant dans lequel il y a deux entrepôts ($m = 2$) avec des
capacités hebdomadaires $a_1 = 1200$ et $a_2 = 1100$. Il y a trois magasins ($n = 3$), les demandes hebdomadaires correspondantes sont $b_1 = 500$, $b_2 = 600$ et $b_3 = 1000$. Voici les coûts d'expédition par unité estimés :


|            | Magasin 1 | Magasin 2 | Magasin 3 |
|:-----------|----------:|----------:|----------:|
| Entrepôt 1 |       300 |       200 |       100 |
| Entrepôt 2 |       250 |       400 |        80 |


Écrivez le modèle PL pour ce cas spécifique.

**CORRECTION :**


Pour l'instance spécifique, on obtient le PL suivant :
$$
\begin{align}
& \min 300 x_{11} + 200 x_{12} + 100 x_{13} + 250 x_{21} + 400 x_{22} + 80 x_{23}\\
& x_{11} + x_{12} + x_{13} \le 1200\\
& x_{21} + x_{22} + x_{23} \le 1100\\
& x_{11} + x_{21}  \ge 500\\
& x_{12} + x_{22}  \ge 600\\
& x_{13} + x_{23}  \ge 1000\\
& x_{11}, x_{12}, x_{13}, x_{21}, x_{22}, x_{23} \ge 0
\end{align}
$$

```python
from mip import *
```

```python
costs = [
    [300, 200, 100],
    [250, 400, 80]
]

capacities = [1200, 1100]

demands = [ 500, 600, 1000]


```

```python
##################
#   Correction   #
##################


def distribution(costs, capacities, demands):

    I = range(len(capacities))
    J = range(len(demands))
    
    m = Model()
    m.verbose = 0
    x = [ [m.add_var() for j in J] for i in I]
    
    m.objective = minimize(xsum(costs[i][j] * x[i][j] for i in I for j in J))
    
    # For each warehouse
    for i in I:
      m += xsum(x[i][j] for j in J) <= capacities[i]
    
    # For each shop
    for j in J:
      m += xsum(x[i][j] for i in I) >= demands[j]
    
    
    # m.write('model.lp')
    status = m.optimize()
    print("status =", status)
    
    if status == OptimizationStatus.OPTIMAL:
      print(f'optimal solution cost: {m.objective_value} euros')
      sol = [
          [x[i][j].x for j in J] for i in I
      ]
      for i in I:
        for j in J:
          if sol[i][j] > 0:
            print(f"- warehouse W{i+1} sends {sol[i][j]} units to shop S{j+1}")

distribution(costs, capacities, demands)
```
